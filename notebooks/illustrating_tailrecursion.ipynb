# Notebook to explore tail recursion with Python

Tail recursion does not exist natively in Python, but I'm using the language because it is the most readable to me. The aim of this notebook is to get to an understanding of it.

Problems where we perform an action several times can take an **iterative approach**, where we repeat the same action over an iterable:

```python

def factorial(n):
    fact = 1
    for n in range(2, n+1):
        fact *= n
    return fact
```

Or we can take a **recursive approach**, where we define a base case, and call the function on itself:
```python

def factorial(n):
    if n < 2:
        return 1
    else:
        return n * factorial(n-1)

```
