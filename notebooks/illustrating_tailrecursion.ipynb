# Notebook to explore tail recursion with Python

Tail recursion does not exist natively in Python, but I'm using the language because it is the most readable to me. The aim of this notebook is to get to an understanding of it.

```python
def factorial(n):
    fact = 1
    for n in range(2, n+1):
        fact *= n
    return fact
```

```python
print(factorial(1))
```
